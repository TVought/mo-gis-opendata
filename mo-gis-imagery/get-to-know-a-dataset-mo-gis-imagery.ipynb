### Q: How have you organized your dataset? Help us understand the key prefix structure of your S3 bucket.

Due to the large number ofr files typically generated by aerial imagery collections, MSDIS organizes data as it was delivered by the vendor. The data shared through AWS are organized as follows:

2023 Northern Missouri Aerial Imagery
    - README
    - Project Level: 2023_NorthernMissouriImagery
        - County Level: 2023_COUNTYNAME_Tiles
            - GeoTIFF Level: GeoTIFF_Tiles
                - Individual GeoTIFF tiles named according to their index location
                - Individual GeoTIFF OVR pyramid files named according to their index location
            - Index Level: Index
                - County imagery index: 2023_COUNTYNAME_shp
            - MrSID Level: MrSID_Tiles
                - Individual MrSID tiles named according to their index location


2024 Southern Missouri Aerial Imagery
    - README
    - Project Level: 2024_SouthernMissouriImagery
        - County Level: 2024_COUNTYNAME_Tiles
            - GeoTIFF Level: GeoTIFF
                - Individual GeoTIFF tiles named according to their index location
                - Individual GeoTIFF OVR pyramid files named according to their index location
            - Index Level: Index
                - County imagery index: 2023_COUNTYNAME_shp
            - MrSID Level: MrSID
                - Individual MrSID tiles named according to their index location

Access to reliable, single-source-of-truth imagery at 6-inch resolution allows GIS professionals to digitize road centerlines with complete confidence that their outputs will be consistent with those of neighboring administrative units.
